```python
import matplotlib.pyplot as plt
import numpy as np

regions = ['North', 'South', 'East', 'West']
q1 = [100, 120, 140, 160]
q2 = [110, 130, 150, 170]

x = np.arange(len(regions))
width = 0.35

plt.bar(x - width/2, q1, width, label='Q1')
plt.bar(x + width/2, q2, width, label='Q2')

plt.xticks(x, regions)
plt.title("Quarterly Sales by Region")
plt.xlabel("Regions")
plt.ylabel("Sales")
plt.legend()
plt.show()
```
