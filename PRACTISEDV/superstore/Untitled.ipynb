```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("superstore.csv")
print(df.head())
```

```python
sales = df.groupby("Category")["Sales"].sum()

print(sales)
```

```python
print(df.columns)
```

```python
sales.plot(kind = "bar")
plt.xlabel("category")
plt.ylabel("Sales")
plt.title("the title")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("superstore.csv")

sales = df.groupby("Category")["Sales"].sum()

plt.pie(sales, labels=sales.index, autopct="%1.1f%%")

plt.title("Sales Distribution by Category")

plt.show()
```
