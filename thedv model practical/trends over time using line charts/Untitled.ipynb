```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("AAPL.CSV")
```

```python
print(df.isnull().sum())
print(df)
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df = df.sort_values('Date')
```

```python
plt.figure(figsize = (12,5))
plt.plot(df['Date'], df['Close'],marker ='o',color ='blue')
plt.grid(True)
plt.title("Line chart")
plt.show()
```
